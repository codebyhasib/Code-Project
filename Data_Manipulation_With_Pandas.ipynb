```python
climate['year'] = climate.index.year
```

```python
climate['month'] = climate.index.month
```

```python
climate['day'] = climate.index.day
```

```python
climate['day_name'] = climate.index.day_name()
```

```python
monthly = climate['meantemp'].resample('ME').mean()
monthly
```

```python
yearly = climate['meantemp'].resample('YE').mean()
yearly.plot(kind = 'bar')
```

```python
monthly_stat = climate['meantemp'].resample('ME').agg(['max','min','mean'])
monthly_stat
```

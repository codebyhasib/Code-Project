```python
#Convert totalcharges to numeric
df['TotalCharges']=pd.to_numeric(df['TotalCharges'],errors='coerce')
df.dtypes
```

```python
#check uniques values of Totalcharges
df['TotalCharges'].unique()
```

```python
#Remove missing values
df=df.dropna()
```

```python
df.isnull().sum()
```

```python
#Check duplicate rows
df.duplicated().sum()
```

```python
#Remove duplicate
df=df.drop_duplicates()
```

```python
df.duplicated().sum()
```

```python
#Checking uniques values of all columns
for col in df.columns:
  print("column name:",col)
  print(df[col].unique())
  print("_" * 50)
```

```python
#Check uniques values of target column
df['Churn'].unique()
```

```python
#Convert churn column into Numeric
df['Churn']=df['Churn'].map({'Yes':1,'No':0})

```

```python
df['Churn'].head()
```

```python
df.dtypes
```

```python
#Check number of uniques values
df.nunique()
```

```python
#Remove unnecessary columns
df=df.drop(['customerID'],axis=1)
```

```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns
```

```python
df=pd.read_csv("/content/cleaned telco customer data.csv")
df
```

```python
df.head()
```

```python
df.shape
```

```python
df.info()
```

```python
df.columns
```

```python
df['TenureGroup']=pd.cut(df['tenure'],bins=[0,12,24,48,72],
 labels=['0-1 Year','1-2 Year','2-4 Year','4-6 Year',])
```

```python
df['TenureGroup'].value_counts()
```

```python
sns.countplot(x='TenureGroup', data=df)
plt.title('Tenure Group Counts')
plt.show()
```

```python
df['MonthlyChargescategory']=pd.cut(df['MonthlyCharges'],bins=[0,35,70,120],
 labels=['low','Medium','High',])
```

```python
df['MonthlyChargescategory'].value_counts()
```

```python
sns.countplot(x='MonthlyChargescategory', data=df)
plt.title('Monthly Charges Category Counts')
plt.show()
```

```python
df['TotalChargescategory']=pd.cut(df['TotalCharges'],bins=[0,1000,3000,9000],
                                   labels=['low','Medium','High',])
```

```python
df['TotalChargescategory'].value_counts()
```

```python
sns.countplot(x='TotalChargescategory', data=df)
plt.title('Total Charges Category Counts')
plt.show()
```

```python
df['AverageMonthlySpend']=df['TotalCharges']/(df['tenure']+1)
```

```python
df['AverageMonthlySpend'].head()
```

```python
df['EngagementScore']=df['tenure'] + df['MonthlyCharges']
```

```python
df['EngagementScore'].head()
```

```python
service_columns = [
    'PhoneService',
    'MultipleLines',
    'InternetService',
    'OnlineSecurity',
    'OnlineBackup',
    'DeviceProtection',
    'TechSupport',
    'StreamingTV',
    'StreamingMovies'
]
df['ServiceCount']=(df[service_columns]=='No').sum(axis=1)
```

```python
df['ServiceCount'].head()
```

```python
sns.histplot(data=df, x='ServiceCount', bins=10)
plt.title('Service Count Distribution')
plt.show()
```

```python
df['LongTermCustomer']=np.where(df['tenure']>=24,1,0)
```

```python
df['LongTermCustomer'].value_counts()
```

```python
df['LongTermCustomer']=np.where(df['tenure']>=24,'No','Yes')
```

```python
df['LongTermCustomer'].value_counts()
```

```python
df['HighMonthlyCharges']=np.where(df['MonthlyCharges']>df['MonthlyCharges'].median(),1,0)
```

```python
df['HighMonthlyCharges'].value_counts()
```

```python
df['PaperlessBilling'] = df['PaperlessBilling'].map({
    'Yes': 1,
    'No': 0
})
```

```python
df['PaperlessBilling'].head()
```

```python
df['SeniorCitizen'] = df['SeniorCitizen'].map({
    1: 'Yes',
    0: 'No'
})
```

```python
df['SeniorCitizen'].head()
```

```python
df['SeniorCitizen'].value_counts()
```

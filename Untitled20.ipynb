```python
score = 82

if score >= 90:
    grade = "A"
elif score >= 80:
    grade = "B"
elif score >= 70:
    grade = "C"
elif score >= 60:
    grade = "D"
else:
    grade = "F"

print(f"Score: {score} | Grade: {grade}")
```

```python
def add_numbers(a, b=10):
    return a + b

# Testing with 2 arguments
result1 = add_numbers(5, 15)
print("Result with (5, 15):", result1)

# Testing with 1 argument (b defaults to 10)
result2 = add_numbers(5)
print("Result with (5):", result2)
```

```python
def find_larger(x, y):
    if x >= y:
        return x
    else:
        return y

larger = find_larger(25, 40)
print("The larger number is:", larger)
```

```python
fruits = ["apple", "banana", "cherry"]
print("Initial list:", fruits)

# 1. Append
fruits.append("orange")

# 2. Modify
fruits[1] = "mango"

# 3. Remove
fruits.remove("cherry")

# 4. Print updated list and length
print("Updated list:", fruits)
print("List length :", len(fruits))
```

```python
product = {"name": "Laptop", "price": 1200, "brand": "Dell"}

# 1. Access
print("Product Name :", product["name"])
print("Product Price:", product["price"])

# 2. Add new key
product["in_stock"] = True

# 3. Update price
product["price"] = 1150

# 4. Iterate and print
print("--- Product Details ---")
for key, value in product.items():
    print(f"{key}: {value}")
```

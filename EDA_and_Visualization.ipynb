```python
class_count = titanic['Pclass'].value_counts().sort_index()
class_label = ['1st class', '2nd class', '3rd class']
# titanic['Pclass'].value_counts().sort_index()

plt.pie(class_count,
        labels = class_label,
        colors = ['red','green','blue'],
       autopct = "%0.2f%%",
       wedgeprops = {'width' : 0.5})
plt.title('Year wise stock price graph', loc = 'center')
plt.show()
```

```python
data = titanic['Age'].dropna()

plt.figure(figsize = (8, 5))
plt.hist(data,
        bins = 15,
        color = 'g',
        alpha = 0.75,
        edgecolor = 'r',
        cumulative = False)

plt.title('Age histplot', loc = 'center')
plt.xlabel('Age')
plt.ylabel('Number of passenger')
plt.show()
```

```python
plt.scatter(tips['total_bill'],tips['tip'])
```

```python
fig, axes = plt.subplots(2,2, figsize = (12, 5))

axes[0,0].scatter(tips['total_bill'],tips['tip'])
axes[0,0].set_title('abcd')
axes[0,0].set_xlabel('total_bill')


axes[0,1].hist(tips['total_bill'])
axes[0,1].set_title('defg')

avg_tip = tips.groupby('day')['tip'].mean()
axes[1,0].bar(avg_tip.index, avg_tip.values)

axes[1,1].hist(data)

plt.savefig("subplot.png")
plt.show()
```

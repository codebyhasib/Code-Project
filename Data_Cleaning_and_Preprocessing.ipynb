```python
import pandas as pd

data = {
    "Name": [
        "  John Smith",
        "MARY JOHNSON  ",
        "Robert@Brown",
        "  alice smith  ",
        "David Jones!!!",
        "SARAH WILSON"
    ],
    "Email": [
        "john@gmail.com",
        " MARY@YAHOO.COM ",
        "robert@gmail.com",
        "alice@gmail.com",
        "david@outlook.com",
        "sarah@gmail.com"    ],
    "City": [
        "New York",
        "new york",
        " NEW YORK ",
        "London",
        "london ",
        "LONDON"
    ],
    "Job": [
        " Data Scientist ",
        "data scientist",
        "DATA SCIENTIST",
        "Software Engineer",
        "?",
        " software engineer "    ],
    "Age": [25, 30, 28, 35, 40, 29]
}

df = pd.DataFrame(data)

print(df)
```

```python
df["Name"] = df["Name"].str.strip()
df['Name']
```

```python
df["City"] = df["City"].str.lower()
df["City"]
```

```python
df["Name"] = df["Name"].str.replace(r"[^\w\s]", "", regex=True)
df["Name"]
```

```python
df["Job"] = df["Job"].replace("?", pd.NA)
df["Job"]
```

```python
df[df["Job"].str.contains("data", case=False, na=False)]
```

```python
df["Name_Length"] = df["Name"].str.len()
df["Name_Length"]
```

```python
df["Email_Domain"] = df["Email"].str.split("@").str[-1]
df["Email_Domain"]
```

```python
import zipfile
import pandas as pd

with zipfile.ZipFile("/content/Superstore.xlsx.zip") as z:

    excel_file_name = [name for name in z.namelist() if name.endswith('.xlsx')][0]
    with z.open(excel_file_name) as f:
        store = pd.read_excel(f)

store
```

```python
print("--- Exact Column Names ---")
print(store.columns.tolist())
print("\n" + "="*50 + "\n")
```

```python
region_sales = store.groupby('Region')['Sales'].sum()
```

```python
print("--- Step 1: Simple GroupBy (Region wise Total Sales) ---")
print(region_sales)
print("\n" + "="*50 + "\n")
```

```python
region_stats = store.groupby('Region')['Sales'].agg(['sum', 'mean', 'count'])
```

```python
print("--- Step 2: Multiple Calculations (sum, mean, count) ---")
print(region_stats)
print("\n" + "="*50 + "\n")
```

```python
region_category_sales = store.groupby(['Region', 'Category'])['Sales'].sum()

print("--- Step 3: Grouping by Two Columns (Region & Category) ---")
print(region_category_sales)
print("\n" + "="*50 + "\n")
```

```python
formatted_table = region_category_sales.unstack()

print("--- Step 4: Unstacked Table (Easier to read) ---")
print(formatted_table)
```

```python
region_metrics = store.groupby('Region').agg(
    Average_Sales=('Sales', 'mean'),
    Average_Profit=('Profit', 'mean'),
    Maximum_Profit=('Profit', 'max')
).reset_index()
```

```python
region_metrics['Average_Sales'] = region_metrics['Average_Sales'].map('${:,.2f}'.format)
region_metrics['Average_Profit'] = region_metrics['Average_Profit'].map('${:,.2f}'.format)
region_metrics['Maximum_Profit'] = region_metrics['Maximum_Profit'].map('${:,.2f}'.format)

print(region_metrics)
```

```python
region_sales_summary = store.groupby('Region')['Sales'].agg(['mean', 'min', 'max', 'sum'])
```

```python
region_sales_summary = region_sales_summary.applymap('${:,.2f}'.format)

print(region_sales_summary)
```

```python
region_sales_named = store.groupby('Region').agg(
    Mean_Sales=('Sales', 'mean'),
    Min_Sales=('Sales', 'min'),
    Max_Sales=('Sales', 'max'),
    Total_Sales=('Sales', 'sum')
)

print(region_sales_named)
```

```python
import matplotlib.pyplot as plt
import pandas as pd
import seaborn as sns
```

```python
sns.set_theme(style="whitegrid")
```

```python
iris = pd.read_csv("/content/iris.csv")
titanic = pd.read_csv("/content/Titanic-Dataset.csv")
```

```python
plt.figure(figsize=(8, 5))
sns.scatterplot(
    data=iris,
    x="sepal_length",
    y="petal_length",
    hue="species",
    style="species",
    size="petal_width",
    sizes=(20, 200),
    palette="Set1",
)
plt.title(
    "1. Multivariate Scatter Plot: Sepal Length vs Petal Length (Hue, Style, Size)"
)
plt.tight_layout()
plt.show()
```

```python
g2 = sns.pairplot(iris, hue="species", palette="Set1")
g2.fig.suptitle("2. Pair Plot: Iris Dataset Matrix", y=1.02)
plt.show()
```

```python
plt.figure(figsize=(7, 5))
numeric_cols = iris.drop(columns=["species"])
corr_matrix = numeric_cols.corr()

sns.heatmap(corr_matrix, annot=True, cmap="coolwarm", fmt=".2f", linewidths=0.5)
plt.title("3. Heatmap: Correlation Matrix of Iris Numerical Features")
plt.tight_layout()
plt.show()
```

```python
g4 = sns.FacetGrid(
    titanic,
    col="Pclass",
    row="Sex",
    hue="Survived",
    palette="Set1",
    height=3,
    aspect=1.2,
)
g4.map_dataframe(sns.scatterplot, x="Age", y="Fare")
g4.add_legend()
g4.fig.suptitle("4. FacetGrid: Age vs Fare by Pclass, Sex, & Survival", y=1.03)
plt.show()
```

```python
g5 = sns.PairGrid(iris, hue="species", palette="Set2")
g5.map_diag(sns.kdeplot, fill=True)
g5.map_offdiag(sns.scatterplot)
g5.add_legend()
g5.fig.suptitle(
    "5. PairGrid: Custom Plots on Diagonal (KDE) & Off-Diagonal (Scatter)",
    y=1.02,
)
plt.show()
```

```python
g6 = sns.jointplot(
    data=iris,
    x="sepal_length",
    y="petal_length",
    hue="species",
    palette="Set1",
    height=6,
)
g6.fig.suptitle(
    "6. Multivariate Joint Plot: Sepal Length vs Petal Length by Species",
    y=1.02,
)
plt.show()
```

```python
plt.figure(figsize=(8, 5))
sns.boxenplot(data=titanic, x="Pclass", y="Age", hue="Sex", palette="Set2")
plt.title("7. Multivariate Boxen Plot: Age Distribution by Pclass and Sex")
plt.tight_layout()
plt.show()
```
